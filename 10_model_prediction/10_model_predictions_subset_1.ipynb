#**Model Predictions**
This notebook focuses on generating predictions on whether a compound in the dataset has antiviral activity using our final models for each target. The dataset was originally too large, so it was fragmented into 10 subsets to facilitate the prediction process, which were finally concatenated. This notebook serves as an example of predictions using all 21 models for each selected target.
